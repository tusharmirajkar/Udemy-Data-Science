```python
import sqlite3
```

```python
connection = sqlite3.connect('example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
cursor.execute('''
Create Table If Not Exists employees(
id Integer Primary Key,
name Text Not Null,
age Integer,
department text
)
''')

connection.commit
```

```python
cursor.execute('''
Insert Into employees(name, age, department)
values('Krish', 32, 'Data Scientist')'''
)
cursor.execute('''
Insert Into employees(name, age, department)
values('Prachu', 82, 'Data Scientist')'''
)
cursor.execute('''
Insert Into employees(name, age, department)
values('Tushar', 18, 'Data Scientist')'''
)

connection.commit()
```

```python
cursor.execute('Select * From employees')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute('''
Update employees
set age = 34
where name = "Tushar"
''')
```

```python
cursor.execute(
    '''
Delete from employees
where name = "Krish"
'''
)
```

```python
connection.close()
```
