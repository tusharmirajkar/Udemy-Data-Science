```python
import matplotlib.pyplot as plt
```

```python
x = [1, 3, 5,6, 7, 9, 12]
y = [1, 4, 7, 8 ,9 , 13, 23]

plt.plot(x, y, color='red', linestyle='--', marker = 'o')
```

```python
x = [1, 3, 9, 10, 18 ,23]
y1 = [2, 4, 5, 6, 78, 9]
y2 = [2, 4, 5, 78, 8, 9]

plt.figure(figsize=(9,5))

plt.subplot(2, 2, 1)
plt.plot(x,y1)
plt.title('plot1')

plt.subplot(2, 2, 2)
plt.plot(x, y2, color = 'red')

plt.subplot(2, 2, 3)
plt.plot(y1, x, color = 'yellow')

plt.subplot(2, 2, 4)
plt.plot(y2, x, color = 'green')

plt.show()
```
